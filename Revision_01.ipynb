### 2. Types of Machine Learning:
- Supervised Learning: Machine which learns from Labeled data.(X-features, y-target)
- UnSupervised Learning: Machine which learns from unlabeled data.(X-features, no target, data clustering)

### 3. Regression & Classification

- Classification: predicts the categorical data.{(X) ---> (y)target is categorical}

### 6. Metrics for Regression
- MAE(mean average error): summation |actual - predict| / no. of errors
- MSE(mean square error): summation (|actual - predict|)^2 / no. of errors
- RMSE: sqrt(MSE)
- R^2:  which tells us the variance in target

### 7. Metrics for Classification
- Confusion matrix: TP, TN, FP, FN
  - Accuracy: Out of all how many correct? (TP+TN)/total possibilities
  - Precision: When the model says POSITIVE, how often is it actually positive?
  - Recall: Out of all the actual positive cases, how many did my model successfully find?
  - F1-score: F1 is useful when you want a balance between Precision and Recall.
  - ---------Actual
  - ---------1  0
  - Pred 1---TP      FP
  - Pred 0---FN      TN

### 8. Overfitting vs Underfitting
- Training and Test Performance poor(high bias): Underfitting
- model learns training data too specifically: Overfitting
