```python
numerical_cols = [
    "age",
    "balance",
    "day",
    "duration",
    "campaign",
    "pdays",
    "previous"
]

categorical_cols = [
    "job",
    "marital",
    "education",
    "default",
    "housing",
    "loan",
    "contact",
    "month",
    "poutcome"
]

target = "y"
```

```python
df.duplicated().sum()
```

```python
counts = df["y"].value_counts()

percentages = df["y"].value_counts(normalize=True) * 100

print(counts)
print(percentages)
```

```python
df[numerical_cols].describe()
```

```python
df["balance"].sort_values(ascending=False).head(10)
```

```python
Q1 = df["balance"].quantile(0.25)
Q3 = df["balance"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
print(Q1, Q3)
print(lower, upper)
```

```python
df[df["balance"] > upper]["balance"].count()
```

```python
df[df["balance"] > upper]["y"].value_counts(normalize=True) * 100
```

```python
df["y"].value_counts(normalize=True) * 100
```

```python
df[categorical_cols].nunique()
```

```python
for col in categorical_cols:
    print(f"---{col}---")
    print(df[col].value_counts())
```

```python
df.groupby("poutcome")["y"].value_counts(normalize=True) * 100
```

Target:

    age → mild right-skew
    other numeric features → need preprocessing consideration

Categorical:
    rare categories → default=yes, month=dec, etc.
    unknown categories → contact, poutcome, job, education
    poutcome → potentially strong relationship with target
